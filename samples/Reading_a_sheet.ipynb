## Reading a sheet

```python
from cirro import DataPortal

portal = DataPortal()
project = portal.get_project_by_name("Test Project")
```

#### Finding a sheet

```python
sheet = project.get_sheet_by_name("Example Samples")
print(sheet)
```

#### Inspecting the columns

```python
for column in sheet.columns:
    print(column.name, column.data_type)
```

#### Reading rows into a DataFrame

`to_dataframe` pages through the whole sheet. The first column, `_row_id`, identifies each row for `update_rows` and `delete_rows`.

```python
df = sheet.to_dataframe()
df.head()
```

#### Reading part of a sheet

Sort the rows and stop after a set number of them. To filter rows, see `Querying_sheets.ipynb`.

```python
from cirro_api_client.v1.models import SqlSortOrder

sheet.to_dataframe(max_rows=3, order_by="read_count", order=SqlSortOrder.DESC)
```
