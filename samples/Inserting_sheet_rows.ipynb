## Inserting rows into a sheet

Each row is a mapping of column name to value, and columns left out of a row are null.

```python
from cirro import DataPortal

portal = DataPortal()
project = portal.get_project_by_name("Test Project")
```

```python
sheet = project.get_sheet_by_name("Example Samples")
```

#### Inserting a few rows

```python
inserted = sheet.insert_rows([
    {"sample_id": "S1", "tissue": "liver", "condition": "control", "read_count": 21000000},
    {"sample_id": "S2", "tissue": "liver", "condition": "treated", "read_count": 19500000},
    {"sample_id": "S3", "tissue": "lung", "condition": "control"}
])
print(f"Inserted {inserted} rows")
```

#### Inserting rows from a DataFrame

`to_dict(orient="records")` turns each row of a DataFrame into the mapping `insert_rows` expects.

```python
import pandas as pd

df = pd.DataFrame({
    "sample_id": ["S4", "S5"],
    "tissue": ["kidney", "lung"],
    "condition": ["treated", "treated"],
    "read_count": [18000000, 22500000]
})

inserted = sheet.insert_rows(df.to_dict(orient="records"))
print(f"Inserted {inserted} rows")
```
