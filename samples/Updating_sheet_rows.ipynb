## Updating rows in a sheet

Rows are identified by `_row_id`, which `to_dataframe` returns as the first column. Read the sheet, find the rows to change, then pass their `_row_id`s to `update_rows` or `delete_rows`.

```python
from cirro import DataPortal

portal = DataPortal()
project = portal.get_project_by_name("Test Project")
```

```python
sheet = project.get_sheet_by_name("Example Samples")
```

#### Finding the rows to change

```python
df = sheet.to_dataframe()
df
```

#### Updating one row

Only the columns you name are written; the rest of the row is left alone.

```python
row_id = int(df.loc[df["sample_id"] == "S3", "_row_id"].iloc[0])

sheet.update_rows({row_id: {"read_count": 20100000}})
```

#### Updating several rows

```python
# Mark every lung sample as treated
lung = df.loc[df["tissue"] == "lung", "_row_id"]

updated = sheet.update_rows({int(row_id): {"condition": "treated"} for row_id in lung})
print(f"Updated {updated} rows")
```

#### Deleting rows

```python
to_delete = df.loc[df["sample_id"] == "S5", "_row_id"]

deleted = sheet.delete_rows([int(row_id) for row_id in to_delete])
print(f"Deleted {deleted} rows")
```
