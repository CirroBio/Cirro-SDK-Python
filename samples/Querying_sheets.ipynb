## Querying sheets with SQL

`query_sheets` runs SQL across all the sheets in a project and returns the result as a DataFrame. Refer to a sheet by its `table_name` (here `example_samples`), not its display name.

```python
from cirro import DataPortal

portal = DataPortal()
project = portal.get_project_by_name("Test project 1")
```

#### Filtering rows

```python
project.query_sheets("""
    SELECT sample_id, condition, read_count
    FROM example_samples
    WHERE tissue = 'liver'
""")
```

#### Summarizing rows

```python
project.query_sheets("""
    SELECT condition, COUNT(*) AS samples, AVG(read_count) AS mean_reads
    FROM example_samples
    GROUP BY condition
""")
```

#### Limiting a large result

`query_sheets` reads the whole result unless `max_rows` stops it early.

```python
project.query_sheets("SELECT * FROM example_samples", max_rows=2)
```
