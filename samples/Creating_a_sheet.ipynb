## Creating a sheet

Sheets are tables of structured data held alongside a project. A `TABLE` sheet holds rows you write to it directly, while a `VIEW` sheet is materialized from a query over other sheets.

This notebook creates the `Example Samples` table used by the other Sheets samples.

```python
from cirro import DataPortal

portal = DataPortal()
project = portal.get_project_by_name("Test project 1")
```

#### Listing the sheets in a project

```python
print(project.list_sheets().description())
```

#### Defining the columns

Each column has a name and a data type. A `required` column cannot be null, and a `unique` column rejects duplicate values.

```python
from cirro_api_client.v1.models import ColumnDataType, ColumnDef

columns = [
    ColumnDef(name="sample_id", data_type=ColumnDataType.STRING, required=True, unique=True),
    ColumnDef(name="tissue", data_type=ColumnDataType.STRING),
    ColumnDef(name="condition", data_type=ColumnDataType.STRING),
    ColumnDef(name="read_count", data_type=ColumnDataType.BIGINT)
]
```

#### Creating the sheet

`sheet_type` says whether the sheet is a `TABLE` or a `VIEW`. `name` is what the sheet is called in Cirro, and `table_name` is how SQL queries refer to it. `namespace_name` groups related tables and cannot be changed after the sheet is created.

```python
from cirro_api_client.v1.models import SheetCreationMode, TableSheetInput

sheet = project.create_sheet(
    TableSheetInput(
        sheet_type="TABLE",
        name="Example Samples",
        description="Sample metadata used by the Sheets examples",
        namespace_name="default",
        table_name="example_samples",
        sheet_creation_mode=SheetCreationMode.STANDARD,
        columns=columns
    )
)
print(sheet)
```
